```python
import pandas as pd
```

```python
file_path = r"C:\Users\HP\Documents\Financial_Data.csv.xlsx"
```

```python
df = pd.read_excel(file_path)
```

```python
# Print all column names to see the exact names
print(df.columns.tolist())
```

```python
# Strip leading/trailing spaces from column names
df.columns = df.columns.str.strip()

# Print column names again to ensure they are cleaned up
print(df.columns.tolist())
```

```python
columns_to_convert = [
    'Total Revenue',  # Ensure this matches the exact column name
    'Net Income',
    'Total Assets',
    'Total Liabilities',
    'Operating Cash Flow'  # Update this as well if the column name is different
]

# Convert the columns to numeric values
for col in columns_to_convert:
    df[col] = pd.to_numeric(df[col], errors='coerce')
```

```python
df.sort_values(by=['Company', 'Fiscal Year'], ascending=[True, False], inplace=True)

df.head()
```

```python
# Calculate YoY % growth
df['Revenue Growth (%)'] = df.groupby('Company')['Total Revenue'].pct_change(-1) * 100
df['Net Income Growth (%)'] = df.groupby('Company')['Net Income'].pct_change(-1) * 100
df['Assets Growth (%)'] = df.groupby('Company')['Total Assets'].pct_change(-1) * 100
df['Liabilities Growth (%)'] = df.groupby('Company')['Total Liabilities'].pct_change(-1) * 100
df['Cash Flow Growth (%)'] = df.groupby('Company')[' Operating Cash Flow'].pct_change(-1) * 100

# View updated DataFrame
df.head(10)
```
